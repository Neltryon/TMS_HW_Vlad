2. Даны матрицы A B C
Найти: А+В, В-А, A*C, A*B*C (руками и кодом)

```python
import numpy as np

A = np.array([
    [1, 2, 4],
    [7, 5, -2],
    [-2, 1, 3]
])

B = np.array([
    [-1, 0, 3],
    [2, -6, 1],
    [-7, 1, 5]
])

C = np.array([
    [-2, 3],
    [4, 0],
    [7, -1]
])

A_plus_B  = A + B
B_minus_A = B - A
A_times_C = A @ C
A_B_C     = A @ B @ C

print("***|A + B|***\n", A_plus_B, "\n")
print("***|B - A|***\n", B_minus_A, "\n")
print("***|A * C|***\n", A_times_C, "\n")
print("***|A * B * C|***\n", A_B_C, "\n")
```
